```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ---------------------------------------------------------
# 1. Modelos e hiperparâmetros
# ---------------------------------------------------------

modelos = {

    "Logistic Regression": (
        LogisticRegression(max_iter=1000),
        {
            "C": [0.01, 0.1, 1, 10, 100],
            "class_weight": [None, "balanced"]
        }
    ),

    "Decision Tree": (
        DecisionTreeClassifier(random_state=42),
        {
            "max_depth": [3, 5, 8, 12, None],
            "min_samples_split": [2, 5, 10, 20]
        }
    ),
"SVM": (
    SVC(probability=True, random_state=42),
    [
        {
            "C": [0.1, 1, 10, 100],
            "kernel": ["rbf"],
            "gamma": ["scale", "auto", 0.01, 0.1, 1]
        }
    ]
),

    "Random Forest": (
        RandomForestClassifier(random_state=42, n_jobs=-1),
        {
            "n_estimators": [100, 200, 500],
            "max_depth": [None, 5, 10, 20]
        }
    ),

    "KNN": (
        KNeighborsClassifier(),
        {
            "n_neighbors": [3, 5, 7, 11, 15],
            "weights": ["uniform", "distance"]
        }
    )
}


# ---------------------------------------------------------
# 2. Validação cruzada
# ---------------------------------------------------------

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ---------------------------------------------------------
# 3. Treinamento + GridSearch
# ---------------------------------------------------------

resultados = []
modelos_treinados = {}

for nome, (modelo, parametros) in modelos.items():

    grid = GridSearchCV(
        estimator=modelo,
        param_grid=parametros,
        cv=cv,
        scoring="roc_auc",
        n_jobs=-1
    )

    grid.fit(X_train_transform, y_train)

    # Melhor modelo encontrado
    melhor_modelo = grid.best_estimator_

    # Previsões no conjunto de teste
    y_pred = melhor_modelo.predict(X_test_transform)
    y_prob = melhor_modelo.predict_proba(X_test_transform)[:, 1]

    # Métricas
    resultados.append({
        "Modelo": nome,
        "Melhores parâmetros": grid.best_params_,
        "CV ROC-AUC": grid.best_score_,
        "Test Accuracy": accuracy_score(y_test, y_pred),
        "Test Precision": precision_score(y_test, y_pred),
        "Test Recall": recall_score(y_test, y_pred),
        "Test F1": f1_score(y_test, y_pred),
        "Test ROC-AUC": roc_auc_score(y_test, y_prob)
    })

    modelos_treinados[nome] = grid


# ---------------------------------------------------------
# 4. Resultado
# ---------------------------------------------------------

resultados_df = pd.DataFrame(resultados)

resultados_df = resultados_df.sort_values(
    "Test ROC-AUC",
    ascending=False
)

resultados_df
```
